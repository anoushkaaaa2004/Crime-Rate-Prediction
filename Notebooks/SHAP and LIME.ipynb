```python
import pandas as pd

# SHAP table
df_shap = pd.DataFrame([
    ["Dataset 1", "GNB", "After SHAP", 0.946, 0.886, 1.000, 0.939],
    ["Dataset 1", "MLP", "After SHAP", 0.993, 1.000, 0.984, 0.992],
    ["Dataset 2", "GNB", "After SHAP", 0.494, 0.489, 0.444, 0.466],
    ["Dataset 2", "MLP", "After SHAP", 0.505, 0.502, 0.541, 0.520]
], columns=["Dataset","Model","Stage","Accuracy","Precision","Recall","F1"])

df_shap["Method"] = "SHAP"

# LIME table
df_lime = pd.DataFrame([
    ["Dataset 1", "GNB", "After LIME", 0.495, 0.491, 0.438, 0.463],
    ["Dataset 1", "MLP", "After LIME", 0.507, 0.503, 0.523, 0.513],
    ["Dataset 2", "GNB", "After LIME", 0.495, 0.491, 0.438, 0.463],
    ["Dataset 2", "MLP", "After LIME", 0.507, 0.503, 0.523, 0.513]
], columns=["Dataset","Model","Stage","Accuracy","Precision","Recall","F1"])

df_lime["Method"] = "LIME"

# Combine
df_final = pd.concat([df_shap, df_lime], ignore_index=True)
# Display table
df_final.style.background_gradient(cmap="Blues")
```

```python
import matplotlib.pyplot as plt

mlp_data = df_final[df_final["Model"] == "MLP"]

plt.figure()

for dataset in ["Dataset 1", "Dataset 2"]:
    subset = mlp_data[mlp_data["Dataset"] == dataset]
    plt.plot(subset["Method"], subset["Accuracy"], marker='o', label=dataset)

plt.title("SHAP vs LIME Comparison (MLP)")
plt.ylabel("Accuracy")
plt.legend()
plt.show()
```

SHAP outperforms LIME in both datasets, especially for MLP, due to better and more stable feature selection.
